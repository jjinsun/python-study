### 04. 사용자 입력(input)
- input 함수 : 콘솔을 통해 사용자로부터 문자열 형태의 입력을 받는 함수
 - input을 통해 받은 압력은 항상 문자열로 입력됨


```python
# input함수 사용법
# my_input = input()
# input을 통해 받은 입력 → 문자열
# 숫자로 활용할 시 '형변환'을 해줘야 함!!
a = int(input())
b = int(input())
print(a + b)
```

```python
# 실수 계산을 하려면 실수형으로 변환
c = float(input())
d = float(input())
print(c + d)
```

```python
# input의 입력에 대한 설명문 추가
name = input("이름을 입력하세요: ")
print(f"제이름은 {name}입니다.")
```

> 설명문 추가 단축키 : control 엔터 입력
> 입력을 받으면 변수에 저장해줘야 한다.!(중요)
> 코딩이 안 될떼는 ESC로 나가준다.

```python
# 여러 자료 입력하기 
국어, 영어, 수학, 사화 = 100, 80, 90, 85
과일1, 과일2, 과일3 = "수박", "참외", "사과"
```

```python
과일1, 과일2, 과일3 = input()
```

```python
# "수박, 사과, 포도"
# 문자열을 쪼개주는 함수 : split
# split은 구분자를 기준으로 문자열을 쪼개서 리스트를 반환함
# 구분자의 기본값은 공백
# 과일1, 과일2, 과일3 = "수박 사과 참외".split()
# print(과일1)
# print(과일2)
# print(과일3)
과일 = " 수박 사과 참외".split() # 공백을 기준으로 나눠줌
print(과일)
```

```python
# input에 적용
과일1, 과일2, 과일3 = input().split()
print(과일1)
print(과일2)
print(과일3)
```
