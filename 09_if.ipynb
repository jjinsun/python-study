```python
# 실습1. 
# 비가 오는 경우
# 날씨가 비 라면? → 날씨와 비가 같다.
날씨 = input("오늘 날씨를 입력하세요.(비 or 맑음)")
if 날씨 == "비":
    print("우산을 챙기세요☔️")

# 맑은 경우
# 날씨가 맑음 라면? → 날씨와 맑음이 같다.
if 날씨 == "맑음":
    print("썬크림을 바르세요☀️")
```

```python
# if - else 문
# 조건이 참일 때는 if문을 조건이 거짓일 때는 else문을 실행
# else : 아니라면의 의미 → 조건이 필요x, if문과 반드시 같이 나와야 함.
a = int(input())
if a > 10:
    print("a는 10보다 커요")
else:
    print("a는 10보다 작아요") # else는 단독으로 사용 못합.
```

```python
# 실습2. 짝수 홀수 판별하기
a = int(input("정수를 입력해주세요."))
if a % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
```

```python
# if - elif -else 구문
# else : 만약 아니라면
# elif : else if의 약자: 그게 아니라 만약 ~라면
# elif는 조건을 반드시 기록합니다.
# if가 있어야만 사용할 수 있음
score = int(input())
# 90점 이상이면 a
if score >= 90:
    print("a")
# (90점 미만이면)80점 이상이면 b
elif score >= 80:
    print("b")
# 70점 이상이면 c
elif score >= 70:
    print("c")
# 60점 이상이면 d
elif score >= 60:
    print("d")
# 그외에는 f
else:
    print("f")

# ⭐️elif는 하나의 덩어리이다.
```

```python
# 실습3. 나이에 따른 영화 관람 가능 여부
age = int(input("나이를 입력해주세요."))
# 0살 이상 12살이하이면 전체 관람가
if age <= 12:
    print("전체관람가")
# 13살 이상 15살이하이면 12세 이상 관람가
elif age <= 15:
    print("12세 이상 관람가")
# 16살 이상 18살이하이면 15세 이상 관람가
elif age <= 18:
    print("15세 이상 관람가")
# 그외에는 청소년 관한불가 가능
else:
    print("청소년 관람불가 가능")
```

```python
# 실습4. 과제 hour, minute, secomd 구하기
Time = int(input("Time:"))
hour = Time//3600
minute = Time%3600//60
secomd = Time%3600%60

if Time >= 3600:
    print(f"{hour}시 {minute}분 {secomd}초")
elif Time >= 60:
    print(f"{minute}분 {secomd}초")
else:
    print(f"{secomd}초")
```

```python
#. 실습5. 과제 편의점 도시락 구매하기
금액 = int(input("금액"))
상품 = input("구매 할 상품명")

if 금액 < 1500:
    print("금액이 부족합니다.")

elif 금액 < 1500:
    if 상품 == "삼각김밥":
        print("금액이 부족합니다.")
    else:
        print("구매 성공🥳")

elif 금액 < 2500:
    if 상품 == "김밥":
        print("금액이 부족합니다.")
    else:
        print("구매 성공🥳")

elif 금액 < 4000:
    if 상품 == "도시락":
        print("금액이 부족합니다.")
    else:
        print("구매 성공🥳")
else:
    print("구매 성공🥳")
```
