```python
# 파이썬 표준 라이브러리
# math : 수학적 연산에 사용되는 모듈
import math

# 1. 올림/내림
# round : 반올림_파일썬 내장 함수
round(3.141492, 2)

# ceil : 올림, 소수점 지정 안 됨
math.ceil(3.14)

# floor : 내림, 소수점 지정 안 됨
math.floor(3.14)

# 2. 제곱, 제곱근
# pow(x,y) : 제곱_x^y를 반환
math.pow(2,3)

# sqrt(x) : 제곱근 반환
math.sqrt(16)

# 3. 상수
# pi : 원주율
math.pi

# 4. 수학 계산 편의 기능
# factorial(x) : x! 팩토리얼 변환
math.factorial(5)

# gcd(x,y) : 최대 공약수
math.gcd(12,20)

# lcm(x,y) : 최소 공배수
math.lcm(30,40)
```

```python
# 실습2. math 모듈 사용해보기
# 문제1. 실제거리계산 : 좌표 두 점 사이 거리 구하기
import math

# 좌표 입력
x1, y1 = 1, 2
x2, y2 = 3, 4

# 거리 계산 (피타고라스 공식)
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

print(f"두 점 사이의 거리",round(distance, 2))
```

```python
# 문제2. 상품나누기: 최소 공배수와 최대 공약수
import math

a = 18   # 학생 수
b = 24   # 선생님 수

# 최대공약수(GCD)
gcd = math.gcd(a, b)
print("최대공약수(GCD):", gcd)

# 최소공배수(LCM) (Python 3.9+)
lcm = math.lcm(a, b)
print("최소공배수(LCM):", lcm)
```

```python
# random 모듈 : 랜덤 값(난수) 생성시 사용
import random
# 1. 난수 생성
# seed(a) : 난수 발생 초기값(seed) 설정_최초 생성한 난수를 고정해야 할 때 사용
random.seed(42)

# random() : 0이상 1미만의 float 난수 반환
random.random()

# uniform(a, b) : a이상 b이하의 실수 난수 반환
random.uniform(1, 10)

# ranint(a, b) : a이상 b이하의 정수 난수 반환
random.randint(1, 100)

# randrange(start, stop, step) : 범위 안의 정수 난수 반환, 간격 지정 가능
random.randrange(0, 100, 5)
```

```python
# 2.렌덤 선택
fruits = ["apple", "watermellon", "banana", "orange"]

# choice(seq) : 시퀀에서 임의의 요소 1개 변환
random.choice(fruits)

# choices(seq, k) : 시퀀스에서 중복 허용 k개의 요소 리스트를 반환
random.choices(fruits, k=2)

# sample(seq, k) : 시퀀스에서 "중복없이" k개 요소 리스트를 반환
random.sample(fruits, k=2)

# shuffle(seq) : 시퀀스의 요소를 무작위로 섞음 → 원본 시퀀스를 변경
numbers = [1,2,3,4,5]
random.shuffle(numbers)
numbers
```

```python
# 실습4. 가위 바위 보 게임 만들기
options = ['가위', '바위', '보']

computer = random.choice(options)
user = input("가위, 바위, 보 중 하나를 입력하세요: ")

# 승패 판정
if user == computer:
    print(f"컴퓨터👾: {computer}, 사용자😈: {user} → 무승부!😌")
elif (user == '가위' and computer == '보') or \
     (user == '바위' and computer == '가위') or \
     (user == '보' and computer == '바위'):
    print(f"컴퓨터👾: {computer}, 사용자😈: {user} → 사용자 승리😈")
elif user in options:
    print(f"컴퓨터👾: {computer}, 사용자😈: {user} → 컴퓨터 승리!👾")
else:
    print("잘못 입력하셨습니다. '가위✌🏻', '바위✊🏻', '보✋🏻' 중 하나만 입력하세요.")
```

```python
# 실습5. 다음생일까지 남은 날짜 계신
import datetime

# 사용자로부터 생일 입력 받기
birthday_str = input("생일을 입력하세요 (월-일, 예: 07-25): ")
month, day = map(int, birthday_str.split('-'))
print(birthday_str)

# 오늘 날짜 구하기
today = datetime.date.today()

# 올해 생일 날짜 생성
birthday_this_year = datetime.date(2025, month, day)

if birthday_this_year < today:
    # 이미 생일이 지났으면 내년 생일로 계산
    birthday_next = datetime.date(2025 + 1, month, day)
    days_left = (birthday_next - today).days
else:
    # 아직 안 지났으면 올해 생일까지 남은 날짜 계산
    days_left = (birthday_this_year - today).days

print(f"다음 생일까지 남은 일수: {days_left}일")
```

```python
def solution(left, right):
    answer = 0
    for i in range(left, right + 1):
        # i가 제곱수인지 확인
        if int(i ** 0.5) == i ** 0.5:
            answer -= i
        else:
            answer += i
    return answer

print(solution(13, 17))  
print(solution(24, 27))  
```

```python
# time  모듈
# 시간의 측정, 지연, 변환과 같은 시간 관련 기능 제공
import time

# 1. 시간 반환
# time()
# Unix 타임스탬프로 반환(1979,1,1부터 경과 초)
time.time()

# ctime() : 현재 시간을 문자열로 반환
time.ctime()
time.ctime(0) # 기준시로 반환

# strftime(): 원하는 포맷의 문자열로 시간 객체 변환
lt = time.localtime()
formatted = time.strftime("%Y-%m-%d %H:%M:%S", lt)
print(formatted)

# strptime(): 문자열을 struct_time 객체로 변환
parsed = time.strptime(formatted, "%Y-%m-%d %H:%M:%S")
print(parsed)

# 2. 시간 지연
# sleep(seconds) : 지정한 초만큼 프로그램이 일시중지
time.sleep(1)
```

```python
# 시간 측정하기
start = time.time()

for i in range(5):
    print(i)
    time.sleep(1)

end = time.time()
print(f"수행시간 : {end-start: .2f}초")
```

```python
# 실습6. 타자 연습 게임 만들기
import random
import time

words = ["apple", "banana", "computer", "python", "window", "keyboard", "mouse", "monitor", "teacher", "student",
    "school", "laptop", "chicken", "flower", "pencil", "orange", "holiday", "picture", "button", "cloud",
    "river", "coffee", "sandwich", "travel", "bridge", "forest", "market", "planet", "rocket", "summer"]

print("👾타자 연습 게임에 오신 것을 환영합니다!👾")
input("🚕엔터키를 누르면 게임이 시작됩니다.🚕")

n_questions = 10
used_words = set()
score = 0

start_time = time.time()

for i in range(1, n_questions + 1):
    # 출제되지 않은 단어만 무작위 선택
    word = random.choice([w for w in words if w not in used_words]) # w는 "word"의 약자, 반복문/리스트 컴프리헨션에서 임시로 쓰는 변수명일 뿐
    used_words.add(word)

    print(f"\n문제 {i}: {word}") # \n은 New Line(새 줄)의 약자
    attempts = 1
    user_input = input("단어를 입력하시오: ")

    while user_input != word and attempts < 3:  # 3번 안에 못 맞추면 정답을 알려주고 넘어가는 형태
        print(f"❌오타! 다시 도전! ({attempts}/3)")
        attempts += 1
        user_input = input("단어를 입력하시오: ")

    if user_input == word:
        print("⭕️정답입니다!⭕️")
        score += 1
    else:
        print(f"💤기회 소진! 정답은 '{word}'였습니다.💤")

end_time = time.time()
total_time = round(end_time - start_time, 2)
print(f"\n🎉게임 종료! 총 소요 시간: {total_time}초, 총 통과 문제: {score}개🎊🍻")
```
