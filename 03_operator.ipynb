### 03. 연산자
- 연산자(operator) : 값과 값 사이에 연산을 수행하는 기호
- 항(Operand) : 연산의 대상이 되는 값

### 연산자의 종류
1. 산술 연산자
2. 비교 연산자
3. 논리 연산자
4. 대입 연산자

### 산술연산자
- 사칙연산을 포함한 수학 연산(계산) 수행
- +, -, *, /, //, %, ** : 나누기, 몫, 나머지, 거듭제곱연산
- 계산 값이 어떤 특정한 값(숫자)

### 비교연산자
- 두 값을 비교하는 연산 수행
- 계산 결과 Ture 또는 False를 반환
- ==, !=, >, <, >=, <= : 같다, 다르다, 크다, 작다, 크거나 같다, 작거나 같다

### 논리연산자
- 논리에 대한 연산을 수행
- True, False에 대한 연산을 수행
- and,or,not

```python
# 산술 연산자
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 3)
# ⭐️단축키 복사 : ption shit 방향키⬇️
# ⭐️단축키 한번에 선택: ption 선택
```

```python
# 변수와 산술 연산
# 반복되는 항이 있다면 변수로 처리
a = 10
b = 3
print("더하기", a + b)
print("빼기", a - b)
print("곱하기", a * b)
print("나누기", a / b)
print("몫", a // b)
print("나머지", a % b)
print("거듭제곱", a ** b)
# ⭐️단축키 : command d
```

```python
# 변수의 값을 변경
x = 10

# x에 5를 더하고 싶다면?
x = x + 5

# x에서 10을 빼고 싶다면?
x = x - 10

# x에 두배를 해주고 싶다면?
x = x * 2

# 결과값 출력
print(x)
```

왜? : 메모리에는 연산영역이 따로 있다. 냉장고에서 꺼내와서 계산, 
⭐️요리하고 요리한 것을 다시 넣어줌.⭐️
> x애 15를 넣어준 것이다.

```python
# 복합 대입 연산자
변수에 자기 자신에 대한 연산 결과를 다시 저장할 때 사용한다!!
# 변수의 값을 변경 
x = 10

# x에 5를 더하고 싶다면?
x += 5

# x에서 10을 빼고 싶다면?
x -= 10

# x에 두배를 해주고 싶다면?
x *= 2

# x에 5로 나누고 싶다면?
x /= 5

# 결과값 출력
print(x)
```

⭐️⭐️연산자에는 연산자 우선 순위⭐️⭐️
→ 더한 다음 저장한다.!
→ 뺀 다음 저장한다.!
→ 곱한 다음 저장한다.!

```python
# 실습1. 대학생의 용돈 관리
# 1. 시작 용돈은 30만 원입니다.
money = 300000
# 2. 개강첫주에 책을 사느라 8만원을썼습니다.
money -= 80000
# 3. 평일 점심값으로 매일 9천 원씩 5일간 사용했습니다.
money -= 9000 * 5
# 4. 과외 아르바이트를 하며 12만 원을벌었습니다.
money += 120000
# 5. 부모님이 용돈을 더 주셔서 현재 금액의 20%를 추가로 받았습니다.
# money = money + (money * 0.2)
# money = money * (1 + 0.2)
# money = money * 1.2
money *= 1.2
# 6. 하지만 전기요금 등 공과금으로 남은 돈의 1/3이 빠져나갔습니다.
# money = money - (money * 1/3)
# money = money * (1-1/3)
# money = m * 2/3
money *= 2/3

print(f"남은 돈은 {money}입니다.")
```

```python
# 문자열 연산(산술 연산)
# + : 문자열의 결합(연결)
print(100 + 200) # 정수연산 → 계산됨
print("100" + "200") # 문자연산 → 결합됨

# ❌ 문자열과 정수의 복합 더하기 연산은 불가능.
print("100" + 200)
```
